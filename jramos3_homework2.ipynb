## Problem 1

#### 1A. Load the gun violence data set (gun-violence-data_01-2013_03-2018.csv) into a DataFrame.

```python
# Load gun violence data set into a DataFrame

df = pd.read_csv('gun-violence-data_01-2013_03-2018.csv', encoding='utf-8', sep=',')
```

```python
# use .dtypes to report data type for each column

df.dtypes
```

#### 1B. Filter the original data frame to show incidents that occurred in Iowa City, IA. Be sure to not include incidents that occurred in Iowa City, CA. Save results to a new data frame named iowa_city_ia. How many incidents occurred in Iowa City, IA? Show the new data frame.

```python
## create new dataframe and use & operator to include IA only

iowa_city_ia = df[(df['city_or_county'] == 'Iowa City') & (df['state'] == 'Iowa')]

print(f"Incidents in Iowa City, IA: {len(iowa_city_ia)}\n")

# print the new DataFrame

print(iowa_city_ia)
```

#### 1C. Filter the original data frame to show incidents that occurred in both Iowa City (like step b) and Cedar Rapids, IA. Be sure not to include Cedar Rapids, NE.  Save results to a new data frame named corridor_ia. How many incidents occurred in Iowa City and Cedar Rapids, IA? Show the new data frame.

```python
# create new dataframe called corridor_ia and include Cedar Rapids, IA

corridor_ia = df[((df['city_or_county'] == 'Iowa City') & (df['state'] == 'Iowa')) | ((df['city_or_county'] == 'Cedar Rapids') & (df['state'] == 'Iowa'))]

print(f"Incidents in Iowa City and Cedar Rapids, IA: {len(corridor_ia)}\n")

print(corridor_ia)
```

#### 1D. Save all Iowa incidents into a new data frame named iowa_df and save it as a JSON file named iowa_incidents.json. Verify the created JSON file is valid using https://jsonlint.com. 

```python
# create new iowa_df dataframe and filter to all Iowa incidents

iowa_df = df[(df['state'] == 'Iowa')]

# create JSON file named iowa_incidents.json

iowa_df.to_json('iowa_incidents.json', orient='records', lines=False)

print(f'Incidents in Iowa: {len(iowa_df)}')
```

#### 1E. In 2016, which cities and states had the most fatalities? Display the maximum number killed in 2016. Show n_killed, city_or_county, and state columns for records that match the maximum number killed.

```python
# convert date column to datetime from an object

df['date'] = pd.to_datetime(df['date'])

# extract year from datetime format

incidents_2016 = df[df['date'].dt.year == 2016]

# get the sum of all values

aggregated_2016 = incidents_2016.groupby(['city_or_county', 'state'], as_index=False)['n_killed'].sum()

# find the maximum of the values 

max_killed_2016 = aggregated_2016['n_killed'].max()

print(f'max fatalities 2016: {max_killed_2016}')

print('-'*20)

max_killed_records = aggregated_2016[aggregated_2016['n_killed'] == max_killed_2016]

display(max_killed_records[['n_killed', 'city_or_county', 'state']]) 
```

## Problem 2

#### Load the FIFA 21 complete player data set (players_21.csv) into a data frame named fifa_21. 

```python
# read data into data frame

fifa_21 = pd.read_csv('players_21.csv', encoding='utf-8', sep=',')
```

#### 2A. Subset the following player IDs (sofifa_id) into a separate data frame named selected_players. Save only the player id and long name fields into this new data frame.

```python
# select the 6 player ids and make a list

player_ids = [231747, 153079, 162835, 200104, 223848, 208421]

# make sure only the sofifa_id and long name fields are printed

selected_players = fifa_21[fifa_21['sofifa_id'].isin(player_ids)][['sofifa_id', 'long_name']]

print(selected_players)
```

#### 2B. Subset players with a height greater than or equal to 191cm and a weight greater than or equal to 90kg into a separate data frame named big_players. Display the number of players in this new data frame.

```python
# subset players with height and weight

big_players = fifa_21[(fifa_21['height_cm'] >= 191) & (fifa_21['weight_kg'] >= 90)]

print(f"There are {len(big_players)} players who meet the criteria.")
```

#### 2C. What is the age of the oldest player(s)? The youngest? What is the median age and the mean age of all players. Store each in variables and display them. Display results in whole numbers.

```python
# find the youngest age in the dataset

youngest_players = fifa_21['age'].min()

# find the median age and round it to a whole number

median_age = round(fifa_21['age'].median())

#find the mean age and round to a whole number 
mean_age = round(fifa_21['age'].mean())

# find the oldest age in the dataset

oldest_players = fifa_21['age'].max()

print('Youngest player(s):', youngest_players)

print('Median player age:', median_age)

print('Mean player age:', mean_age)

print('Oldest player(s):', oldest_players)
```

#### 2D. Show the long name, age, and height for the oldest player(s) who prefer their left foot.

```python
# find the oldest player that prefers their left foot

left_foot_player = fifa_21[fifa_21['preferred_foot'] == 'Left']['age'].max()

# use & to make sure that the long name, age and height are included 

left_foot_player = fifa_21[(fifa_21['preferred_foot'] == 'Left') & (fifa_21['age'] == left_foot_player)][['long_name', 'age', 'height_cm']]

display(left_foot_player)
```

#### 2E. Display the number of athletes and the number of columns in this data set

```python
# use .nunique() to find the number of distinct values for sofifa_id column

athletes = fifa_21['sofifa_id'].nunique()

# use len(fifa_21.columns) to find the number of variables (columns) in the data set

variables = len(fifa_21.columns)

print('Athletes:', athletes)

print('Variables:', variables)
```
